```python
import yfinance as yf

spy = yf.download("SPY", start="2023-01-01", end="2026-08-06")
spy = spy[["Close"]].reset_index()
spy.columns = ["data", "close"]
spy["data"] = spy["data"].dt.strftime("%Y-%m-%d")

print(spy.head())
print(spy.tail())
```

```python
spy["retorno_pct"] = spy["close"].pct_change() * 100
print(spy.tail(10))

spy.to_csv("C:\Desafio QuantAI/data/spy_precos.csv", index=False)
```
